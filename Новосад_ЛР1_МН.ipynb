## Завдання 2. Скалярний добуток і довжина векторів

Дано вектори:
```python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])
```

1. Обчисліть скалярний добуток двома способами: `np.dot()` і оператором `@`.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами.

````python
u = np.array([2, -3,6])
v = np.array([5, 1,4])

d1 = np.dot(u, v)
d2 = u @ v
n_u = np.linalg.norm(u)
n_v = np.linalg.norm(v)
c_a = d1 / (n_u * n_v)

print("Обчисліть скалярний добуток способом - np.dot:", d1)
print("Обчисліть скалярний добуток оператором @:", d2)
print("Евклідова норма u:", n_u)
print("Евклідова норма v:", n_v)
print("Косинус кута між векторами:", c_a)
````

````python
A = np.array ([
[6.0, 2.0],
[2.0, 3.0]])

values, vectors = np.linalg.eigh(A)

left = A @ vectors
right = vectors * values

print ("Власні значення:", values)
print ("Власні вектори:\n", vectors)
print("nepeBipka:", np.allclose(left, right))
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

X = np.column_stack((np.ones_like(x),x))
coef, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)

a, b = coef
y_pred =X @ coef
sse = np.sum((y -y_pred) ** 2)

print("a =", a)
print("b =", b)
print("Прогноз:", y_pred)
print("Сума квадратів залишків:", sse)
````
